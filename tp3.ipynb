```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 4*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(x, y1, 'b', label='sin(x)', linewidth=2)
ax.plot(x, y2, 'r--', label='cos(x)', linewidth=2)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Fonctions trigonométriques')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

```python
n_points = 100
x = np.random.rand(n_points)
y = 2*x + 1 + np.random.randn(n_points) * 0.5
colors = np.random.rand(n_points)
sizes = 100 * np.random.rand(n_points)

fig, ax = plt.subplots(figsize=(8, 6))
scatter = ax.scatter(x, y, c=colors, s=sizes, alpha=0.7, cmap='viridis')
ax.set_xlabel('variable x')
ax.set_ylabel('variable y')
ax.set_title('Nuage de points avec couleurs et tailles aléatoires')
plt.colorbar(scatter, label='Intensité')
plt.show()
```

```python
data1 = np.random.normal(0,1,1000)
fig,(ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.hist(data1, bins=30, color='skyblue', alpha=0.7, edgecolor='black')
ax1.set_title('Histogramme simple')

ax2.hist(data1, bins=30, color='lightcoral', alpha=0.7, edgecolor='black', density=True, cumulative=True)
ax2.set_title('Histogramme cumulatif normalisé')

plt.tight_layout()
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
valeurs1 = [23, 45, 56, 12]
valeurs2 = [30, 35, 40, 20]
x_pos = np.arange(len(categories))

fig, ax = plt.subplots(figsize=(10, 5))
bar_width = 0.4

bars1 = ax.bar(x_pos - bar_width/2, valeurs1, bar_width ,label='Serie 1', color='lightblue', edgecolor='grey')
bars2 = ax.bar(x_pos + bar_width/2, valeurs2, bar_width, label='Serie 2', color='lightcoral', edgecolor='grey')

ax.set_xlabel('Catégories')
ax.set_ylabel('Valeurs')
ax.set_title('Diagramme en barres groupées')
ax.set_xticks(x_pos)
ax.set_xticklabels(categories)
ax.legend()

for bar in bars1 + bars2:
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, height + 1, f'{height}')

plt.show()
```
