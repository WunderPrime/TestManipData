```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

1. Importation du fichier

```python
df = pd.read_excel("data_tp2.xlsx", sheet_name="Simple", na_values=["-", "", " ", "NA", "N/A", "na", "n/a"])
print(df.head(10))
```

2. Colonnes avec données manquantes et types des colonnes

```python
print("Le nombre de valeurs manquantes")
print(df.isnull().sum())
print("Types de données manquantes")
print(df.dtypes)
```

```python
manquants = pd.DataFrame({
    "nb_manquants": df.isna().sum(),
    "type": df.dtypes
})
manquants
```

3. Traitement des données manquantes

```python
for col in ['Open Price','High Price','low Price','last Price']:
    df[col]=df[col].fillna(df[col].median())

print(df.head(10))
```

4. Normalisation de la colonne Volume

```python
vmin, vmax = df["Volume"].min(),df["Volume"].max()
df["Volume Normalisé"] = (df["Volume"] - vmin) / (vmax - vmin)
print(df.head(10))
```

5. Suppression des doublons

```python
print(df.duplicated().sum())
print(df.duplicated())
```

```python
df=df.drop_duplicates()
print(df.duplicated().sum())
```

6. Export au format CSV

```python
df.to_csv("out_data.csv", index=False)
```
