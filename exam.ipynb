```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
'user id': [1, 2, 3, 4, 1, 6, 7, 1, 8],
'age': [25, 32, np.nan, 45, 25, np.nan, 35, 25, 40],
'income': [500.0, 741, 750.0, 800, 500.0, 1100, 1200, 500.0, np.nan],
'click rate': [0.15, 0.22, 0.18, np.nan, 0.15, 0.30, np.nan, 0.15,
0.12]})
```

```python
df[df['age'].isna()] 
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df = df.drop_duplicates()
```

```python
df['income'] = (df['income'] - df['income'].min()) / (df['income'].max()
- df['income'].min())
```

```python
df.to_json('output_df.json', orient='records', indent=4)
```
